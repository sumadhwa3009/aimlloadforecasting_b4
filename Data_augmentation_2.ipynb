```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("cleaned_data_2.csv")
print(df.head())
```

```python
print(df.columns)
```

```python
df['time'] = pd.to_datetime(df['time'])
```

```python
df['temp_aug'] = df['temperature'] + np.random.normal(0, 1, len(df))
```

```python
df['es_load_actual_entsoe_transparency_aug'] = df['es_load_actual_entsoe_transparency'] * (1 + np.random.uniform(-0.05, 0.05, len(df)))
```

```python
df['es_load_actual_entsoe_transparency_lag1'] = df['es_load_actual_entsoe_transparency'].shift(1)
df['es_load_actual_entsoe_transparency_lag2'] = df['es_load_actual_entsoe_transparency'].shift(2)
```

```python
df['es_load_actual_entsoe_transparency_ma3'] = df['es_load_actual_entsoe_transparency'].rolling(3).mean()
```

```python
df = df.dropna()
```

```python
df.to_csv("augmented_data_2.csv", index=False)
```
